### 1. La física

Una esfera que cae en un fluido viscoso llega enseguida a una **velocidad
límite**, cuando el peso, el empuje y el arrastre se equilibran. Con arrastre
de Stokes ($F = 6\pi\eta r v$, válido a bajo número de Reynolds):

$$v_{\lim} = \frac{2}{9}\,\frac{(\rho_e - \rho_f)\,g}{\eta}\,r^2$$

Es decir: **$v_{\lim}$ contra $r^2$ es una recta por el origen**, y de su
pendiente sale $\eta$.

Dos condiciones de validez que hay que verificar, no suponer:

- **Número de Reynolds.** $\mathrm{Re} = \rho_f\,v\,(2r)/\eta$. Stokes vale
  para $\mathrm{Re} \ll 1$. Con las esferas más grandes se puede violar, y
  el apartamiento aparece en los residuos. Ésa es la versión de esta clase
  del "¿hasta dónde?" de la Clase 5.
- **Efecto de pared.** El tubo no es infinito. La corrección de Ladenburg,
  a primer orden, es
  $v_{\infty} \approx v_{\text{medida}}\,(1 + 2{,}104\,r/R_{\text{tubo}})$.
  Con $r/R \sim 0{,}1$ ya es una corrección del 20 %: **no es despreciable**,
  y no corregirla es un sistemático que te va a aparecer como una viscosidad
  demasiado alta.
